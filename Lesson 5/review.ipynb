```python
def is_palindrome(word):
    word = str(word)
    return word[::-1] == word
word = input("enter a word: ")
print(is_palindrome(word))
```

```python
def check_number(number):
    if number > 0:
        return "positive"
    elif number < 0:
        return "negative"
    else:
        return "zero"
number = int(input("enter a number: "))
print(check_number(number))
```

```python
def is_polindrome(number):
    original = number
    reverse = 0
    while number > 0:
        reverse = reverse * 10 + number % 10
        number = number // 10
    return original == reverse
number = int(input("enter a number: "))
print(is_polindrome(number))
```

```python
def check(func):
    def wrapper(x):
        if x<0:
            return "You must enter positive numbers"
        else:
             return func(x)

    return wrapper

@check
def sqrt(x):
    return x**(1/2)
print(sqrt(-16))
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        try:
            return func(*args, **kwargs)
        except ZeroDivisionError:
            return "Denominator can't be zero."
    return wrapper

@decorator
def div(a,b):
    return a/b
print(div(0,-1))
```
