Input: "a!1b@2c"

```python
# Q2
people = [
    {"ism": "Ali", "yosh": 20, "imtiyoz": False},
    {"ism": "Vali", "yosh": 65, "imtiyoz": False},
    {"ism": "Guli", "yosh": 18, "imtiyoz": True}
]

first = []
second = []

for p in people:
    if p["yosh"] >= 60 or p["imtiyoz"]:
        first.append(p["ism"])
    else:
        second.append(p["ism"])

result = first + second

print(" ".join(result))
```

```python
# Q5
def reverse_letters_digits(s):

    s = list(s)

    left = 0
    right = len(s) - 1

    while left < right:

        if not s[left].isalnum():
            left += 1

        elif not s[right].isalnum():
            right -= 1

        else:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1

    return "".join(s)


print(reverse_letters_digits("a!1b@2c"))
```
