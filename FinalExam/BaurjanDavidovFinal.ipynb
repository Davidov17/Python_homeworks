 You have a dataset (customer_orders.csv) containing information about customer orders. The dataset has the following columns:

 OrderID: Unique identifier for each order.
 CustomerID: Unique identifier for each customer.
 Product: Name of the product ordered.
 Quantity: Number of units ordered.
 Price: Price per unit.

 Tasks:

 1)Group the data by CustomerID and filter out customers who have made less than 20 orders.
 2)Identify customers who have ordered products with an average price per unit greater than $120.

```python
# Task 2
def check_brackets(s):
    brackets = {'(': ')', '[': ']', '{': '}'}
    stack = []
    for char in s:
        if char in brackets.values():
            stack.append(char)
        elif char in brackets.keys():
            if stack and stack[-1] == brackets[char]:
                stack.pop()
            else:
                return False
check_brackets("()")
```

```python
# Task3
def shift_list(nums, k):
    return nums[k+1:] + nums[:k+1]
shift_list([1,2,3,4,5,6,7], 3)
```

```python
# task 5
import pandas as pd
df = pd.read_csv('customer_orders.csv')
# 1st task
df['Total_orders'] = df.groupby('CustomerID')['Quantity'].transform('sum')
customers_more_orders = df[df['Total_orders'] > 20]
customers_more_orders
# 2nd task
df['Average_price'] = df.groupby('CustomerID')['Price'].transform('mean')
customer_higher_price = df[df['Average_price'] > 120]
customer_higher_price
```

```python
# Task 4
import pandas as pd
recipe = pd.DataFrame({
    "recipe_name": ["hotdog","hotdog","pilov","pilov","pilov","pizza","pizza","pizza"],
    "ingredient_id": [1,2,3,4,5,6,7,8],
    "ingredient_name": ["sausage","bread","rice","carrot","meat","flour","tomato","cheese"]
})

fridge = pd.DataFrame({
    "ingredient_id": [1,2,3,4,6],
    "ingredient_name": ["sausage","bread","rice","carrot","flour"]
})

recipe_ingredients = recipe.groupby("recipe_name")["ingredient_id"].apply(set)
fridge_ingredients = set(fridge["ingredient_id"])

available_recipes = [name for name, ingredients in recipe_ingredients.items() if
                     ingredients.issubset(fridge_ingredients)]

print(available_recipes)
```
