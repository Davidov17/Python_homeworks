```python
import csv

from jedi.cache import memoize_method


class BookNotFoundException(Exception):
    pass
class BookAlreadyBorrowedException(Exception):
    pass
class MemberLimitExceededException(Exception):
    pass

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_borrowed = False
    def __str__(self):
        if self.is_borrowed:
            status = "Borrowed"
        else:
            status = "Available"
        return f"{self.title} by {self.author} ({status})"
class Member:
    Max_books = 3

    def __init__(self, name):
        self.name = name
        self.borrowed_books = []
    def borrow_book(self, book):
        if len(self.borrowed_books) >= self.Max_books:
            raise MemberLimitExceededException(
                f"{self.name} can not borrow more than {self.Max_books} books"
            )
class Library:
    def __init__(self, name, books):
        self.name = name
        self.books = []
        self.members = []
    def add_book(self, book):
        self.books.append(book)
    def add_member(self, member):
        self.name.append(member)

    def find_book(self, title):
        for book in self.books:
            if book.title == title:
                return book
        raise BookNotFoundException(f"Book {title} not found")
    def borrow_book(self, member, title):
        book = self.find_book(title)

        if book.is_borrowed:
            raise BookAlreadyBorrowedException(
                f"Book {title} already borrowed by {member.name}"
            )
    def return_book(self, member, title):
        book = self.find_book(title)

        if book in member.borrowed_books:
            member.return_book(book)
            book.is_borrowed = False
            print(f"{member.name} returned {title}")
        else:
            print(f"{title} not borrowed by {member.name}")

library = Library()

b1 = Book("1984", "George Orwell")
b2 = Book("Python Basics", "John Doe")
b3 = Book("Data Science", "Jane Smith")
b4 = Book("Algorithms", "CLRS")

library.add_book(b1)
library.add_book(b2)
library.add_book(b3)
library.add_book(b4)

m1 = Member("Alice")
library.add_member(m1)


try:
    library.borrow_book(m1, "1984")
    library.borrow_book(m1, "Python Basics")
    library.borrow_book(m1, "Data Science")
except Exception as e:
    print(e)

try:
    library.borrow_book(m1, "Algorithms")
except Exception as e:
    print("Exception:", e)

try:
    library.borrow_book(m1, "1984")
except Exception as e:
    print("Exception:", e)


try:
    library.borrow_book(m1, "Unknown Book")
except Exception as e:
    print("Exception:", e)

library.return_book(m1, "1984")

try:
    library.borrow_book(m1, "Algorithms")
except Exception as e:
    print("Exception:", e)




```

```python
import csv
headers = ["Name", "Subject", "Grade"]
data = ["Alice", "Math", 85], ["Bob", "Science", 78], ["Carol", "Math", 92], ["Dave", "History", 74]

with open("grades.csv", "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(headers)
    writer.writerows(data)

```

```python
import csv
totals = {}
counts = {}
with open("grades.csv", "r", newline="") as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        subject = row["Subject"]
        grade  = float(row["Grade"])



        totals[subject] = totals.get(subject, 0) + grade
        counts[subject] = counts.get(subject, 0) + 1

averages = {subject: totals[subject]/ counts[subject] for subject in totals}

for subject, avg in averages.items():
    print(f"{subject}: {avg:.2f}")

with open("average_grades.csv", "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Subject", "Average"])

    for subject, avg in averages.items():
        writer.writerow([subject, avg])
print("average_grades.csv created")
```
