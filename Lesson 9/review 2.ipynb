Reviewing CSV and JSON files

```python
import csv

headers = ["Name","Subject","Grade"]
columns = [
    ['Alice','Math',85],
   ['Bob','Science',78],
   ['Carol','Math',92],
   ['Dave','History',74]
]

with open('marks.csv', 'w', newline="") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(headers)
    writer.writerows(columns)

with open('marks.csv', 'r') as csv_file:
    reader = csv.reader(csv_file)
    for row in reader:
        print(row)

```

```python
import csv
subjects = {}
with open('marks.csv', 'r') as csv_file:
    reader = csv.DictReader(csv_file)
    for row in reader:
        subject = row['Subject']
        grade = int(row['Grade'])
        if subject not in subjects:
            subjects[subject] = []
        subjects[subject].append(grade)

averages = {}

for subject, grades in subjects.items():
    averages[subject] = sum(grades)/len(grades)

print(averages)

heads = ["Subject", "Average Grade"]
with open('averages.csv', 'w', newline="") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(heads)
    writer.writerows(averages)
    with open('averages.csv', 'r') as csv_file:
        reader = csv.reader(csv_file)
        for row in reader:
            print(row)
```
