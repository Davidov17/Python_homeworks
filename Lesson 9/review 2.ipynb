Reviewing CSV and JSON files

```python
import csv

headers = ["Name","Subject","Grade"]
columns = [
    ['Alice','Math',85],
   ['Bob','Science',78],
   ['Carol','Math',92],
   ['Dave','History',74]
]

with open('marks.csv', 'w', newline="") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(headers)
    writer.writerows(columns)

with open('marks.csv', 'r') as csv_file:
    reader = csv.reader(csv_file)
    for row in reader:
        print(row)

```

```python
import csv
subjects = {}
with open('marks.csv', 'r') as csv_file:
    reader = csv.DictReader(csv_file)
    for row in reader:
        subject = row['Subject']
        grade = int(row['Grade'])
        if subject not in subjects:
            subjects[subject] = []
        subjects[subject].append(grade)

averages = {}

for subject, grades in subjects.items():
    averages[subject] = sum(grades)/len(grades)

print(averages)

heads = ["Subject", "Average Grade"]
with open('averages.csv', 'w', newline="") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(heads)
    writer.writerows(averages)
    with open('averages.csv', 'r') as csv_file:
        reader = csv.reader(csv_file)
        for row in reader:
            print(row)
```

```python
import json
import csv
tasks = [
    {
        "ID": 1,
        "Task": "Do laundry",
        "Completed": False,
        "Priority": 3
    },
    {
        "ID": 2,
        "Task": "Buy groceries",
        "Completed": True,
        "Priority": 2
    },
    {
        "ID": 3,
        "Task": "Finish homework",
        "Completed": False,
        "Priority": 1
    }
]

with open('tasks.json', 'w') as json_file:
    json.dump(tasks, json_file, indent=4)
with open('tasks.json', 'r') as json_file:
    data = json.load(json_file)
with open('tasks.csv', 'w', newline="") as csv_file:
    if len(data)>0:
        writer = csv.DictWriter(csv_file, fieldnames=data[0].keys())
        writer.writeheader()
        writer.writerows(data)
with open('tasks.csv', 'r') as csv_file:
    reader = csv.reader(csv_file)
    for row in reader:
        print(row)
```
