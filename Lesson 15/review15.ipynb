```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.core.completer import position_to_cursor

```

```python
x = np.arange(-10, 10)
y = x**2-4*x+4
plt.plot(x,y, color='red')
plt.title('Square of numbers')
plt.xlabel('x')
plt.ylabel('y', rotation=0)
plt.show()
```

```python
x = np.arange(-np.pi, np.pi, 0.01)
y = np.sin(x)
z = np.cos(x)
plt.plot(y, color='blue', label='sin(x)')
plt.plot(z, color='red', label='cos(x)')
plt.legend()
plt.show()
```

```python
x = np.arange(-10, 10, 0.1)
y1 = x**3
y2 = np.sin(x)
y3 = np.e**x

plt.subplot(2,2,1)
plt.plot(y1, color='blue', label='x^3')
plt.legend()
plt.subplot(2,2,2)
plt.plot(y2, color='yellow', label='sin(x)')
plt.legend()
plt.subplot(2,2,3)
plt.plot(y3, color='green', label='e^x')
plt.legend()
plt.subplot(2,2,4)
x = np.arange(0, 100)
y4 = np.log(x+1)
plt.plot(y4, color='red', label='log(x+1)')
plt.legend()
plt.title('Functions', loc='left')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
import random
x = np.random.randint(0, 10,10)
y = np.random.randint(0, 10,10)
sizes = np.abs(np.random.randn(10))*100

plt.scatter(x, y, s=sizes, c=sizes,  cmap='viridis')
plt.title('Scatter', loc='left')
plt.xlabel('x')
plt.ylabel('y')
plt.colorbar()
plt.grid(True)
plt.show()
```

```python
data = np.random.randn(1000)

plt.hist(data, bins=30, color='blue', edgecolor='black', alpha=0.5)
plt.title('Histogram', loc='center')
plt.xlabel('x')
plt.ylabel('y', rotation=0)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure()
ax = plt.axes(projection='3d')
x = np.arange(-5, 5, 0.01)
y = np.arange(-5, 5, 0.01)
X, Y = np.meshgrid(x, y)
z = np.cos(np.sqrt(X**2+Y**2))
ax.plot_surface(X, Y, z)
plt.show()




```

```python
categories = ['Product A', 'Product B', 'Product C', 'Product D', 'Product E']
colors = ['red', 'green', 'blue', 'yellow', 'magenta']
sales_values = [200, 150, 250, 175, 225]
plt.bar(categories, sales_values, color=[c.lower() for c in colors], width=0.5, align='center')
plt.title('Bar Chart')
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()
```

```python
time_periods = ['T1', 'T2', 'T3', 'T4']
x = np.arange(len(time_periods))

# Sample data for each category
category_a = [10, 15, 20, 25]
category_b = [5, 10, 15, 10]
category_c = [8, 12, 10, 15]

plt.bar(x, category_a, label= 'Data set1', color='blue')
plt.bar(x, category_b, bottom=category_a, label= 'Data set2', color='green')
plt.bar(x, category_c, bottom= np.array(category_a)+np.array(category_b), label= 'Data set3', color='yellow')
plt.legend()
plt.title('Bar Chart')
plt.xlabel('Period')
plt.ylabel('Time')
plt.xticks(x, time_periods)
plt.show()
```
