```python
word = "abcabcdabcdeabcdefabcdefg"
vowels = "aeiou"
result = ""
count = 0
for i in range(len(word)):
    result += word[i]
    count += 1
    if count == 3 and i != len(word) - 1:
        if word[i] in vowels:
            continue
        result += "_"
        count = 0

print(result)
```

```python
import random
MAX_ATTEMPTS = 10
attempts = 0
random_integer = random.randint(1,100)
while attempts < MAX_ATTEMPTS:
    num = int(input("Guess a number between 1 and 100: "))
    attempts += 1
    if num<1 or num>100:
        print("Enter a number between 1 and 100")
        continue
    if num > random_integer:
        print("Too high!")
    elif num < random_integer:
        print("Too low!")
    else:
        print("You guessed right in", attempts, " attempts")
        break
else:
    print("you lost!, Number was:", random_integer)

    choice = input("Do you want to play again? (y/n): ")
    if choice not in ("Y", "y", "yes", "YES", "ok"):
        print("Game over")



```

```python
password = input("Enter a password: ")
if len(password) < 8:
    print("Password is too short!")
if not any(char.isupper() for char in password):
    print("Password must contain an uppercase letter")
else:
    print("Password is strong")
```

```python
import random
choices = ["rock", "paper", "scissors"]
computer_score = 0
player_score = 0
while computer_score < 5 and player_score < 5:
    player = input("Rock, Paper, or Scissors: ").lower()
    computer = random.choice(choices)
    print("Computer chose:", computer)

    if computer == player:
        print("It's a tie!")
    if (computer == "rock" and player == "paper") or (computer == "paper" and player == "scissors") or (computer == "scissors" and player == "rock"):
        print("you win!")
        player_score += 1
    else:
        print("computer wins!")
        computer_score += 1
    print("computer score:", computer_score, "player score:", player_score)
print("Game over!")



```
