```python
class Student:
    def __init__(self, age):
        self._age = age
    @property
    def age(self):
        return self._age
    @age.setter
    def age(self, value):
        if value<0:
            print("Age can not be less than 0")
        else:
            self._age = value
S = Student(18)
print(S.age)
```

```python
# Lesson 8 Task 1
class Animal:
    def __init__(self, name, age, weight, color):
        self.name = name
        self.age = age
        self.weight = weight
        self.color = color
    def walk(self):
        print("Walking")

    def eat(self):
        print("Eating")

    def __str__(self):
        return f"{self.name}, {self.age}, {self.weight}, {self.color}"
p_animal = Animal("Masha", 6, 120, "blue")
print(p_animal)
p_animal.walk()
p_animal.eat()

class Child1(Animal):
    def __init__(self, name, age, weight, color, width):
        super().__init__(name, age, weight, color)
        self.width = width
    def sleep(self):
        print("Ch1 is Sleeping")
    def __str__(self):
        return f"{self.name}, {self.age}, {self.weight}, {self.color}, {self.width}"

Ch1 = Child1("Sasha", 2, 60, "blue", 1)
print(Ch1)
Ch1.sleep()
Ch1.walk()

class Child2(Animal):
    def __init__(self, name, age, weight, color, height):
        super().__init__(name, age, weight, color)
        self.height = height

    def drink(self):
        print("Child2 is Drinking")

    def __str__(self):
        return f"{self.name}, {self.age}, {self.weight}, {self.color}, {self.height}"
Ch2 = Child2("Misha", 1.5, 55, "light blue", 1.1)
print(Ch2)
Ch2.drink()

class Child3(Animal):
    def __init__(self, name, age, weight, color, voice):
        super().__init__(name, age, weight, color)
        self.voice = voice

    def lay(self):             
        print("Child3 is laying")

    def __str__(self):
        return f"{self.name}, {self.age}, {self.weight}, {self.color}, {self.voice}"

Ch3 = Child3("Kasha", 1.4, 50, "dark blue", "moo")
print(Ch3)
Ch3.lay()





```

```python
import os

class Account:
    def __init__(self, account_number, name, balance):
        self.account_number = account_number
        self.name = name
        self.balance = balance
    def __str__(self):
        return f"{self.account_number}, {self.name}, {self.balance}"
class Bank:
    def __init__(self):
        self.accounts = {}
        self.load_from_file()

    def create_account(self, name, initial_deposit):
        account_number = len(self.accounts)+1
        account = Account(account_number, name, initial_deposit)
        self.accounts[account_number] = account
        self.save_to_file()
        print("Account created successfully")
        print(account)


    def view_account(self, account_number):
        account = self.accounts.get(account_number)
        if account:
            print(account)
        else:
            print("Account not found")

    def deposit(self, account_number, amount):
        account = self.accounts.get(account_number)
        if account and amount > 0:
            account.balance += amount
            self.save_to_file()
            print("Account deposited successfully")
        else:
            print("Wrong amount")
    def withdraw(self, account_number, amount):
        account = self.accounts.get(account_number)
        if account and 0 < amount <= account.balance:
            account.balance -= amount
            self.save_to_file()
            print("Account withdrawn successfully")
        else:
            print("Wrong amount")
    def save_to_file(self):
        with open("accounts.txt", "w") as file:
            for acc in self.accounts.values():
                file.write(f"{acc.account_number}, {acc.name}, {acc.balance}\n")
    def load_from_file(self):
        if os.path.isfile("accounts.txt"):
            with open("accounts.txt", "r") as file:
                for line in file:
                    account_number, name, balance = line.strip().split(",")
                    self.accounts[int(account_number)] = Account(int(account_number), name, float(balance))
bank = Bank()
while True:
    print("1. Create an account")
    print("2. View an account")
    print("3. Deposit an account")
    print("4. Withdraw an account")
    print("5. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1:
        name = input("Enter your name: ")
        balance = int(input("Enter your balance: "))
        bank.create_account(name, balance)
    elif choice == 2:
        account_number = int(input("Enter your account number: "))
        bank.view_account(account_number)
    elif choice == 3:
        account_number = int(input("Enter your account number: "))
        amount = float(input("Enter your amount: "))
        bank.deposit(account_number, amount)
    elif choice == 4:
        account_number = int(input("Enter your account number: "))
        amount = float(input("Enter your amount: "))
        bank.withdraw(account_number, amount)
    elif choice == 5:
        print("Thank you for your time")
        break
```
