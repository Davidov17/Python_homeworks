```python
import requests
from bs4 import BeautifulSoup
from jedi.inference.arguments import try_iter_content

url = "https://data.worldbank.org/country"
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
countries = {}
sections = soup.find_all('section')
for section in sections:
    title = section.find('h3')
    countries[title.text] = []
    print(title.text)
    names = section.find_all('a')
    for name in names:
        countries[title.text].append(name.text)
        print('\t', name.text)
```

```python
html_doc = """
<h4>5-Day Weather Forecast</h4>
<table>
    <thead>
        <tr>
            <th>Day</th>
            <th>Temperature</th>
            <th>Condition</th>
        </tr>
    </thead>
    <tbody>
        <tr>
            <td>Monday</td>
            <td>25°C</td>
            <td>Sunny</td>
        </tr>
        <tr>
            <td>Tuesday</td>
            <td>22°C</td>
            <td>Cloudy</td>
        </tr>
        <tr>
            <td>Wednesday</td>
            <td>18°C</td>
            <td>Rainy</td>
        </tr>
        <tr>
            <td>Thursday</td>
            <td>20°C</td>
            <td>Partly Cloudy</td>
        </tr>
        <tr>
            <td>Friday</td>
            <td>30°C</td>
            <td>Sunny</td>
        </tr>
    </tbody>
</table>


"""
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html_doc, 'html.parser')
print(soup.prettify())
```

```python
infos = {}
tbody = soup.find('tbody')
rows = tbody.find_all('tr')
for row in rows:
    tds = row.find_all('td')
    title = tds[0].get_text(strip=True)
    infos[title] = []
    for td in tds[1:]:
        infos[title].append(td.get_text(strip=True))
print(infos)
```
