```python
class Date:
    def __init__(self, month, day, year):
        if not (1 <= month <= 12):
            raise ValueError('Invalid month')
        if not (1 <= day <= 31):
            raise ValueError('Invalid day')
        self.month = month
        self.day = day
        self.year = year

    def format_numeric(self):
        return f"{self.month}/{self.day}/{self.year}"

    def format_long(self):
        months = [
            'January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'
        ]
        return f"{months[int(self.month - 1)]} {self.day} {self.year}"
    def format_day_first(self):
        months = [
            'January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'
        ]
        return f"{self.day} {months[int(self.month - 1)]} {self.year}"

    try:
        date = Date(12,25,2025)

        print(format_numeric())
        print(format_long())
        print(format_day_first())

    except:
        print("Error")

```
