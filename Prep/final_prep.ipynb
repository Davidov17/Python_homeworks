```python
def invert_dict(d):
    reversed = {value: key for key, value in d.items()}
    return reversed
invert_dict({'a':1,'b':2,'c':3})
```

```python
def romanToInt(s):
    roman_nums = {"I":1,"V": 5,"X" : 10,"L" : 50,"C" : 100,"D" : 500,"M" : 1000}
    res = 0
    for a, b in zip(s, s[1:]):
        if roman_nums[a] < roman_nums[b]:
            res -= roman_nums[a]
        else:
            res += roman_nums[a]
    return res + roman_nums[s[-1]]
romanToInt("XIV")
```
