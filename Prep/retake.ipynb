```python
people = [
    {"ism": "Ali", "yosh": 20, "imtiyoz": False},
    {"ism": "Vali", "yosh": 65, "imtiyoz": False},
    {"ism": "Guli", "yosh": 18, "imtiyoz": True}
]

new_list = []
for i in range(len(people)):
    if people[i]['yosh']>=65:
        new_list.append(people[i]['ism'])
for j in range(len(people)):
    if people[j]['imtiyoz']==True:
        new_list.append(people[j]['ism'])
new_list
for k in range(len(people)):
    if not people[k]['ism'] in new_list:
        new_list.append(people[k]['ism'])
new_list
```
