```python
import numpy as np
a = np.array([32, 100, 212, 77])
@np.vectorize
def convert(F):
    C = (F-32)*(5/9)
    return C
convert(a)
```

```python
import numpy as np
numbers = np.array([2,3,4,5])
powers = np.array([1,2,3,4])
@np.vectorize
def custom(num, power):
    return num**power
custom(numbers, powers)
```

\begin{cases}
4x + 5y + 6z = 7 \\
3x - y + z = 4 \\
2x + y - 2z = 5
\end{cases}

```python
import numpy as np
W = np.array([
    [4,5,6],
    [3,-1,1],
    [2,1,-2]
])
Y = np.array([7,4,5])
np.linalg.solve(W, Y)
```

\begin{cases}
10I_1 - 2I_2 + 3I_3 = 12 \\
-2I_1 + 8I_2 - I_3 = -5 \\
3I_1 - I_2 + 6I_3 = 15
\end{cases}

```python
import numpy as np
values = np.array([
    [10,-2,3],
    [-2, 8, -1],
    [3, -1, 6]
])
outputs = np.array([
    12,-5,15
])
np.linalg.solve(values, outputs)
```

```python
from PIL import Image
import numpy as np
with Image.open('birds.jpg') as img:
    img_arr = np.array(img)
    print(img.mode)
```

```python
flipped = img_arr[::-1, :,:]
flipped = Image.fromarray(flipped)
flipped.save('flipped.jpg')
flippedhor = img_arr[:,::-1,:]
flippedhor = Image.fromarray(flippedhor)
flippedhor.save('flippedhor.jpg')

with Image.open('flippedhor.jpg') as img:
    display(img)

```

```python
import numpy as np
from PIL import Image
noise_strength =25

noise = np.random.normal(0, noise_strength, img_arr.shape)

noisy = img_arr + noise

noisy = np.clip(noisy, 0, 255).astype(np.uint8)
Image.fromarray(noisy).save('noisy.jpg')

with Image.open('noisy.jpg') as img:
    display(img)
```
