```python
def divide(nums):
    for n in nums:
       try:
           yield 10 / n
       except ZeroDivisionError:
            yield "Division by zero"
for value in divide([2,1,0,5]):
    print(value)
```

```python
FILE_NAME = 'sample.txt'
def add_employee(emp_id, name, position, salary):
    with open(FILE_NAME, 'a') as f:
        record = f"{emp_id} ,{name}, {position}, {salary}\n"
        f.write(record)
        print("Employee date added successfully!\n")

def view_employees():
    with open(FILE_NAME, 'r') as f:
        print(f.read())


def search_employee(emp_id):
    found = False
    try:
        with open(FILE_NAME, 'r') as f:
            for line in f:
                data = line.strip().split(',')
                if data[0] == emp_id:
                    print("employee found")
                    print(f"ID: {data[0]}")
                    print(f"Name: {data[1]}")
                    print(f"Position: {data[2]}")
                    print(f"Salary: {data[3]}")
                    found = True
                    break
        if not found:
            print("employee not found")
    except FileNotFoundError:
        print("employee not found")

def update_info(emp_id):
    updated = False
    new_lines = []
    try:
        with open(FILE_NAME, 'r') as f:
            lines = f.readlines()

        for line in lines:
            data = line.strip().split('')
            if data[0] == emp_id:
                print("\nEmployeefound, please enter new details: ")
                name = input("Enter a new name: ")
                position = input("Enter a new position: ")
                salary = input("Enter a new salary: ")
                new_line = f"{emp_id} {name} {position} {salary}"
                new_lines.append(new_line)
                updated = True
            else:
                new_line.append(line)
        if updated:
            with open(FILE_NAME, 'w') as f:
                f.writelines(new_lines)
            print("updated successfully")
        else:
            print("employee not found")

    except FileNotFoundError:
        print("employee not found")

def delete_employee(emp_id):
    new_lines = []
    with open(FILE_NAME, 'r') as f:
        lines = f.readlines()
        for line in lines:
            data=line.strip().split(',')
            if data[0] == emp_id:
                continue
            else:
                new_lines.append(line)

    with open(FILE_NAME, 'w') as f:
        f.writelines(new_lines)

while True:
    print("1. add employees")
    print("2. view employees")
    print("3. search employee")
    print("4. update info")
    print("5. delete employee")
    print("6. exit")
    choice = input("Enter choice: ")
    if choice == "1":
        emp_id = input("Enter employee ID: ")
        name = input("Enter employee name: ")
        position = input("Enter employee position: ")
        salary = input("Enter employee salary: ")
        add_employee(emp_id, name, position, salary)
        break
    elif choice == "2":
        view_employees()
        break
    elif choice == "3":
        emp_id = input("Enter employee ID: ")
        search_employee(emp_id)
        break
    elif choice == "4":
        emp_id = input("Enter employee ID: ")
        update_info(emp_id)
        break
    elif choice == "5":
        emp_id = input("Enter employee ID: ")
        delete_employee(emp_id)
        break
    elif choice == "6":
        break
    else:
        print("Invalid choice")
        break
```

```python

import string

def count_word_frequency():
    try:
        with open("sample.txt", "r") as f:
            text = f.read().lower()   # ignore capitalization

        # remove punctuation
        for p in string.punctuation:
            text = text.replace(p, "")

        words = text.split()  # split into words

        freq = {}
        for word in words:
            freq[word] = freq.get(word, 0) + 1

        for word, count in freq.items():
            print(word, ":", count)

    except FileNotFoundError:
        print("sample.txt not found")

count_word_frequency()
```

```python
import string
from collections import Counter
def disp_output():
    try:
        with open("sample.txt", "r") as f:
            data = f.read().lower()
            words = data.split()
            count = len(words)
            print("Total words", ":", count)


            for p in string.punctuation:
                data = data.replace(p, "")
            words1 = data.split()
            freq = Counter(words1)
            value = int(input("Enter a number of common words you want: "))
            print("\nTop", value,  "most common words:")
            for word, count in freq.most_common(value):
                 print(word, "-", count, 'times')
    except FileNotFoundError:
        print("sample.txt not found")
disp_output()
```
