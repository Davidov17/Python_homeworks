```python
columns = df.select_dtypes(include=[object]).columns
for column in columns:
    df[column] = df[column].str.strip()
```

```python
df.fillna('NaN')
```

```python
df['age'] = df['age'].replace('twenty', '20')
df['age'] = df['age'].str.replace('years', '')
df['age']
```

```python
df['age']
```

```python
df['score'] = df['score'].replace('ninety', '90')
df['score'] = df['score'].astype(float)
df['score']
```

```python
df
```

```python
df['gpa'] = df['gpa'].replace('four point five', '4.5')
df['gpa'] = df['gpa'].str.replace(',', '.')
df['gpa'] = df['gpa'].astype(float)
```

```python
df['gpa']
```

```python
df
```

```python
df['attendance'] = df['attendance'].str.strip('%')
df['attendance'] = df['attendance'].astype(float)
```

```python
df['attendance']
```

```python
df
```

```python
df['money_spent'] = df['money_spent'].str.strip('$')
df['money_spent'] = df['money_spent'].str.replace('USD', '')
df['money_spent'] = df['money_spent'].str.replace(',', '.')
df['money_spent'] = df['money_spent'].astype(float)
df['money_spent']
```

```python
df
```

```python
def convert_to(value):
    try:
        return pd.to_datetime(float(value), unit='s')
    except:
        return pd.to_datetime(value)
df['date_of_join'] = df['date_of_join'].apply(convert_to)
df['event_time'] = df['event_time'].apply(convert_to)
```

```python
df
```

```python
df['email'] = df['email'].str.lower().str.strip()
df['email'] = df['email'].str.replace('@@', '@')
df['email'] = df['email'].str.replace('..', '.')
df['email'] = df['email'].str.replace('-', '')
df['email'] = df['email'].str.strip('?')
pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

df.loc[~df["email"].str.match(pattern, na=False), "email"] = None
```

```python
df['email']
```

```python
df
```

```python
df["phone"] = df["phone"].str.replace(r'x\d+', '', regex=True)
df['phone'] = df['phone'].str.replace('[^0-9]', '', regex=True)
df['phone'] = df['phone'].str[-10:]
df.loc[df['phone'].str.len() != 10, 'phone'] = None
df['phone'] = df['phone'].apply(lambda x: f'+1-{x[0:3]}-{x[3:6]}-{x[6:10]}' if pd.notnull(x) else None)


```

```python
df['phone']
```

```python
df
```

```python
df[['addr_city', 'addr_district', 'addr_postal']] = None

df['addr_postal'] = df['address_raw'].str.extract(r'(\b\d{6}\b)')
df['addr_city'] = df['address_raw'].str.extract(r'(Tashkent)')

df['addr_district'] = df['address_raw'].str.extract(r'([A-Za-z\s]+)\s+district')

df.loc[df['addr_district'].isna(), 'addr_district'] = (df['address_raw'].str.extract(r'Tashkent\s+([A-Za-z\s]+)'))

# df = df[~df['address_raw'].str.contains('BROKEN', na=False)]
df.loc[df['address_raw'].str.contains('BROKEN', na=False),
       ['addr_city', 'addr_district', 'addr_postal']] = None
```

```python
df['gender'] = df['gender'].str.lower()
df['gender'] = df['gender'].str.replace('fmale', 'female')
df['gender'] = df['gender'].str.replace('femlae', 'female')
df['gender'] = df['gender'].str.capitalize()
df['gender'] = df['gender'].fillna('Unknown')
df['gender']
```

```python
df
```

```python
df['course'] = df['course'].str.lower()
df['course'] = df['course'].str.replace('ds', 'data science')
df['course'] = df['course'].str.replace('d.s.', 'data science')
df['course'] = df['course'].str.replace('data-sciens', 'data science')
df['course'] = df['course'].str.replace('pythno', 'python')
df['course'] = df['course'].str.replace('pyhton', 'python')
df['course'] = df['course'].str.capitalize()
df['course'] = df['course'].fillna('Other')
df['course']
```

```python
df['status'] = df['status'].str.lower()
```

```python
import json
df['profile_json'] = df['profile_json'].str.replace("'", '"')
df['profile_json'] = df['profile_json'].str.replace('INVALID_JSON_DATA', '')
df['profile_json'] = df['profile_json'].str.replace('{hobbies', '{"hobbies"')

def load_json(x):
    if pd.isna(x) or str(x).strip() == "":
        return None
    try:
        return json.loads(x)
    except:
        return None

df['parsed_json'] = df['profile_json'].apply(load_json)


```

```python
df
```

```python
df_new = pd.json_normalize(
    df[df['parsed_json'].notna()]['parsed_json']
)
```

```python
df_new
```

```python
df_new['hobbies'] = df_new['hobbies'].apply(lambda x: ','.join(x) if isinstance(x, list) else x)
df_new['skills.soft'] = df_new['skills.soft'].apply(lambda x: ','.join(x) if isinstance(x, list) else x)
# df_new['hobbies'] = df_new['hobbies'].astype(str)
# df_new['hobbies'] = df_new['hobbies'].str.strip('[]')
# df_new['hobbies']
```

```python
df_new["devices"] = df_new["devices"].apply(lambda x: x if isinstance(x, list) else [])
dev_slots = df_new["devices"].apply(pd.Series)
dev_cols = []
for i in dev_slots.columns:
    series_i = dev_slots[i].apply(lambda v: v if isinstance(v, dict) else {})
    d = pd.json_normalize(series_i).add_prefix(f"device{i+1}_")
    dev_cols.append(d)
final = pd.concat([df_new.drop(columns=["devices"])] + dev_cols, axis=1)

```

```python
df = pd.concat([df, final], axis=1)
# df_nonnull = df_new[df_new['devices'].notna()]
# df_exploded = df_nonnull.explode('devices')
# df_devices = pd.json_normalize(df_exploded['devices'])
```
