```python
number = int(input("Enter a positive integer number: "))
def is_prime(number):
    if number<=1:
        return False
    for i in range(2, number):
        if number % i == 0:
            return False
        else:
            return True
print(is_prime(number))
```

```python
def generate_primes(limit):
    primes = []
    for n in range(2, limit + 1):
        is_prime = True
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(n)
    return primes


limit = int(input("Yuqori chegara sifatida musbat butun son kiriting: "))
print(generate_primes(limit))
```

```python
def ranged_primes(n,m):
    primes = []
    for i in range(n,m):
        is_prime = True
        for j in range(2, int(i ** 0.5) + 1):
            if i % j == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(i)
    return primes
print(ranged_primes(25,50))
```

```python
example = "a!1b@2c"
lettersnums = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z', '1', '2', '3', '4', '5', '6', '7', '8', '9', '0']
reversed = ''
puncs = {}
for i in range(len(example)):
    if example[i]  in lettersnums:

        reversed += example[i]
    else:
        puncs[i]=example[i]
reversed = reversed[::-1]
new_list=[]
for i in range(len(reversed)):
    new_list.append(reversed[i])
for n in puncs.keys():

    new_list.insert(n, puncs[n])
new_string = ''.join(new_list)
new_string
```
