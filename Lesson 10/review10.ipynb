```python
import requests
CITY = "Tashkent"
API = "https://api.openweathermap.org/data/3.0/onecall?lat={lat}&lon={lon}&exclude={part}&appid={API key}"

params = {
    "q": CITY,
    "appid": API,
    "units": "metric"
}
response = requests.get("https://api.openweathermap.org/data/2.5/weather", params=params)
if response.status_code == 200:
    print(response.json())

    temperature = response.json()["main"]["temp"]
    humidity = response.json()["main"]["humidity"]
    description = response.json()["weather"][0]["description"]

    print(f"Weather in {CITY}: ")
    print(f"Temperature: {temperature} C")
    print(f"Humidity: {humidity} %")
    print(f"Description: {description}")

else:
    print("Error fetching weather data")
    print(response.status_code)
```

```python
import requests
import random

API = "https://developer.themoviedb.org/docs/getting-started/"

genres = {
    "action": 28,
    "comedy": 35,
    "drama": 18,
    "horror": 27,
    "romance": 10749,
    "thriller": 53,
    "sci_fiction": 878
}


user_genre = input("Enter a genre(action, comedy, drama, horror, romance, thriller, sci_fiction): ").lower()
if user_genre not in genres:
    print("Invalid genre")
else:
    params = {
    "api_key": API,
    "with_genres": genres[user_genre],
    "language": "en-US"
    }

response = requests.get("https://api.themoviedb.org/3", params=params)
if response.status_code == 200:
    data = response.json()
    print(data)
    movies = data["results"]


    if movies:
        movie = random.choice(movies)
        print("Movie Recommendation:")
        print("Movie: ", movie["title"])
        print("release_date: ", movie["release_date"],)
        print("Overview: ", movie["overview"])
    else:
        print("No Movie Recommendation")

else:
    print("Error fetching data")
    print(response.status_code)
```
